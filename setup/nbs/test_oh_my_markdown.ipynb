# oh-my-markdown feature tour

TL;DR > One cell per syntax group. Double-click plain text in a cell to see its source; the theme follows VS Code light/dark. <

| Syntax | Result |
|:--|:--|
| `==text==` / `--text--` | yellow / orange highlight |
| `[text]{.blue}` | colored tag, 9 presets |
| `[[Note]]` / `[[Note\|alias]]` | wiki link to `Note.md` |
| `[^id]` | footnote with hover preview |
| `<Ctrl+K>` | keyboard key |
| `> [!TIP+] "Title"` | foldable callout |
| `=== "Tab"` / `??? "Title"` | tabs / collapsible details |
| `:::figure` `:::table` `:::equation` + `@fig:x` | numbered blocks and cross-refs |
| `:::gallery` / `:::timeline` | image grid / timeline |
| `TL;DR > ... <` | summary banner |

## 1. Inline

==yellow highlight==, --orange highlight--, ~~strike~~, **bold**, *italic*, `inline code`

Keys: <Ctrl+Shift+P>, <Cmd+K>, <kbd>Esc</kbd>

Links: [external](https://example.com), [[Design tokens]], [[Design tokens|aliased wiki link]]

Citations: \cite{smith_2024}, \citep{doe_2023, roe_2025}

Footnotes: hover the badge[^a], or this one[^b].

## 2. Styled spans

Write `[text]{.color}` to tag any word or phrase:

[signup]{.green} [checkout]{.pink} [demo]{.yellow} [payment]{.teal} [beta]{.blue} [feature-test]{.purple} [urgent]{.red} [later]{.gray} [draft]{.orange}

Markdown inside works: [**bold** and `code`]{.blue}

Extra attributes: [custom style]{style="font-weight:700; letter-spacing:.08em"}, [hover for title]{.purple title="hello from title"}, [anchor]{#my-anchor .teal}

Not a span (no braces right after `]`): [plain brackets] {.blue}

## 10. Tips

> [!TIP]
> - **Width**: Settings → `omkMarkdown.maxWidth` (default `50em`, `none` for full width).
> - **Editing**: double-click plain text to edit; double-clicks on folds, tabs, buttons and links are ignored.
> - **Copy**: uses the VS Code clipboard and keeps line breaks.
> - **Preview**: the same syntax works in the `.md` Markdown preview.
